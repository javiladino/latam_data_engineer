```python
import urllib.request
import os

# 1. Configuración: ¿Dónde quieres guardar los archivos?
# Esto creará una carpeta llamada 'Datos_Taxi' en tu Google Drive
ruta_destino = '/content/drive/My Drive/Datos_Taxi_2025/'

if not os.path.exists(ruta_destino):
    os.makedirs(ruta_destino)
    print(f"Carpeta creada en: {ruta_destino}")

# 2. El año que queremos procesar
year = "2025"

# 3. Bucle para descargar los meses del 1 al 12
for mes_num in range(1, 13):
    # Formateamos el mes para que sea 01, 02... en lugar de 1, 2...
    mes = f"{mes_num:02d}"
    
    # Construimos la URL y el nombre del archivo
    url = f"https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_{year}-{mes}.parquet"
    nombre_archivo = f"taxi_data_{year}_{mes}.parquet"
    ruta_completa = os.path.join(ruta_destino, nombre_archivo)
    
    print(f"Intentando descargar: {nombre_archivo}...")
    
    try:
        # Descarga directa al Drive
        urllib.request.urlretrieve(url, ruta_completa)
        print(f"✅ Guardado con éxito en tu Drive.")
        
    except Exception as e:
        # Si el archivo no existe en el servidor, nos avisa y sigue con el siguiente
        print(f"❌ No disponible: El mes {mes} aún no tiene datos o la URL es incorrecta.")

print("\n--- Proceso finalizado ---")
```
